# Water demand patterns across SafeSip hubs

Example analysis using the open dataset. This notebook loads
`hub_daily_consumption.csv` and shows:

1. Average daily litres dispensed per hub
2. Uptime distribution
3. Weekly demand pattern

Health planners can extend this to correlate reliable water access with
clinic catchment areas and outreach scheduling.

```python
import pandas as pd

df = pd.read_csv("../data/sample/hub_daily_consumption_sample.csv", parse_dates=["date"])
df.head()
```

## 1. Average daily litres per hub

```python
daily = df.groupby("hub_id")["litres_dispensed"].mean().round(1)
print(daily)
print("\nNetwork daily mean: %.1f litres" % df["litres_dispensed"].mean())
```

## 2. Uptime distribution

```python
print(df["uptime_pct"].describe().round(1))
low_uptime = df[df["uptime_pct"] < 95]
print("\nDays below 95%% uptime: %d of %d" % (len(low_uptime), len(df)))
```

## 3. Weekly demand pattern

```python
df["weekday"] = df["date"].dt.day_name()
weekly = df.groupby("weekday")["litres_dispensed"].mean().round(1)
print(weekly)
```
